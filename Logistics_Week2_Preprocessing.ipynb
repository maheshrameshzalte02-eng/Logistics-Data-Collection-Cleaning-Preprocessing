```python
# ============================================================
# WEEK 2 TASK
# DATA COLLECTION, CLEANING & PREPROCESSING
# LOGISTICS DATA ANALYSIS
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler, StandardScaler

import warnings
warnings.filterwarnings("ignore")

print("=" * 70)
print("WEEK 2 - DATA COLLECTION, CLEANING & PREPROCESSING")
print("=" * 70)


# ============================================================
# 2. FIND DATASET AUTOMATICALLY
# ============================================================

print("\n" + "=" * 70)
print("1. DATA COLLECTION")
print("=" * 70)

dataset_name = "DataCoSupplyChainDataset.csv"

possible_paths = [
    dataset_name,
    os.path.join("data", dataset_name)
]

file_path = None

for path in possible_paths:
    if os.path.exists(path):
        file_path = path
        break

if file_path is None:

    print("❌ DATASET NOT FOUND")
    print("\nCurrent Working Directory:")
    print(os.getcwd())

    print("\nFiles available in current folder:")
    for item in os.listdir():
        print(" -", item)

    raise FileNotFoundError(
        "\nPlease keep DataCoSupplyChainDataset.csv "
        "in the same folder as this notebook."
    )

print("✅ Dataset found!")
print("File location:", file_path)


# ============================================================
# 3. LOAD DATASET
# ============================================================

df = pd.read_csv(
    file_path,
    encoding="latin-1"
)

print("\n✅ Dataset loaded successfully!")

print("Number of Records:", df.shape[0])
print("Number of Columns:", df.shape[1])


# ============================================================
# 4. INITIAL DATA INSPECTION
# ============================================================

print("\n" + "=" * 70)
print("2. INITIAL DATA INSPECTION")
print("=" * 70)

print("\nFirst 5 Records:")
display(df.head())

print("\nLast 5 Records:")
display(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")

for i, column in enumerate(df.columns, 1):
    print(i, "-", column)

print("\nData Types:")
display(df.dtypes)


# ============================================================
# 5. DATASET INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("3. DATASET INFORMATION")
print("=" * 70)

df.info()


# ============================================================
# 6. STATISTICAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("4. STATISTICAL SUMMARY")
print("=" * 70)

display(df.describe())


# ============================================================
# 7. DATA QUALITY CHECK
# ============================================================

print("\n" + "=" * 70)
print("5. DATA QUALITY ASSESSMENT")
print("=" * 70)


# Missing values
missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing_values.index,
    "Missing_Values": missing_values.values,
    "Missing_Percentage": (
        missing_values.values / len(df) * 100
    ).round(2)
})

missing_table = missing_table[
    missing_table["Missing_Values"] > 0
].sort_values(
    "Missing_Values",
    ascending=False
)

print("\nMissing Value Report:")

if len(missing_table) > 0:
    display(missing_table)
else:
    print("✅ No missing values found.")


# ============================================================
# 8. DUPLICATE CHECK
# ============================================================

duplicate_count = df.duplicated().sum()

print("\nDuplicate Records:", duplicate_count)


# ============================================================
# 9. CREATE CLEAN DATASET COPY
# ============================================================

print("\n" + "=" * 70)
print("6. DATA CLEANING")
print("=" * 70)

df_clean = df.copy()

original_rows = df_clean.shape[0]
original_columns = df_clean.shape[1]

print("Original Rows:", original_rows)
print("Original Columns:", original_columns)


# ============================================================
# 10. REMOVE DUPLICATES
# ============================================================

duplicates_before = df_clean.duplicated().sum()

print("\nDuplicate rows before cleaning:")
print(duplicates_before)

df_clean = df_clean.drop_duplicates()

duplicates_after = df_clean.duplicated().sum()

print("Duplicate rows after cleaning:")
print(duplicates_after)

print(
    "Rows removed:",
    duplicates_before - duplicates_after
)


# ============================================================
# 11. MISSING VALUE HANDLING
# ============================================================

print("\n" + "=" * 70)
print("7. MISSING VALUE HANDLING")
print("=" * 70)

missing_before = df_clean.isnull().sum().sum()

print(
    "Total missing values before handling:",
    missing_before
)

# Identify missing-value columns
missing_columns = df_clean.isnull().sum()

missing_columns = missing_columns[
    missing_columns > 0
].sort_values(
    ascending=False
)

if len(missing_columns) > 0:

    print("\nColumns containing missing values:")
    display(missing_columns)

    # Numerical columns -> Median
    numerical_columns = df_clean.select_dtypes(
        include=np.number
    ).columns

    for column in numerical_columns:

        if df_clean[column].isnull().sum() > 0:

            df_clean[column] = df_clean[column].fillna(
                df_clean[column].median()
            )

    # Categorical columns -> Mode
    categorical_columns = df_clean.select_dtypes(
        exclude=np.number
    ).columns

    for column in categorical_columns:

        if df_clean[column].isnull().sum() > 0:

            mode_value = df_clean[column].mode()

            if len(mode_value) > 0:

                df_clean[column] = df_clean[column].fillna(
                    mode_value[0]
                )

else:

    print("✅ No missing values require treatment.")


missing_after = df_clean.isnull().sum().sum()

print(
    "\nTotal missing values after handling:",
    missing_after
)


# ============================================================
# 12. DATE COLUMN PROCESSING
# ============================================================

print("\n" + "=" * 70)
print("8. DATE COLUMN PROCESSING")
print("=" * 70)

date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

for column in date_columns:

    if column in df_clean.columns:

        df_clean[column] = pd.to_datetime(
            df_clean[column],
            errors="coerce"
        )

        print(
            "✅ Converted:",
            column,
            "→ datetime"
        )


# ============================================================
# 13. CHECK INVALID DATES
# ============================================================

print("\nInvalid date values:")

for column in date_columns:

    if column in df_clean.columns:

        invalid_dates = df_clean[column].isnull().sum()

        print(
            column,
            ":",
            invalid_dates
        )


# ============================================================
# 14. DATA TYPE VERIFICATION
# ============================================================

print("\n" + "=" * 70)
print("9. DATA TYPE VERIFICATION")
print("=" * 70)

dtype_table = pd.DataFrame({
    "Column": df_clean.columns,
    "Data_Type": df_clean.dtypes.astype(str).values
})

display(dtype_table)


# ============================================================
# 15. DELIVERY DELAY FEATURE
# ============================================================

print("\n" + "=" * 70)
print("10. FEATURE ENGINEERING")
print("=" * 70)

real_days = "Days for shipping (real)"
scheduled_days = "Days for shipment (scheduled)"

if (
    real_days in df_clean.columns
    and scheduled_days in df_clean.columns
):

    df_clean["Delivery_Delay"] = (
        df_clean[real_days]
        - df_clean[scheduled_days]
    )

    print("✅ Delivery_Delay feature created.")


# ============================================================
# 16. DELIVERY PERFORMANCE
# ============================================================

def classify_delivery(delay):

    if pd.isna(delay):
        return "Unknown"

    elif delay > 0:
        return "Late"

    elif delay == 0:
        return "On Time"

    else:
        return "Early"


if "Delivery_Delay" in df_clean.columns:

    df_clean["Delivery_Performance"] = (
        df_clean["Delivery_Delay"]
        .apply(classify_delivery)
    )

    print(
        "✅ Delivery_Performance feature created."
    )


# ============================================================
# 17. DELIVERY RESULT
# ============================================================

if (
    "Delivery_Delay" in df_clean.columns
    and "Delivery_Performance" in df_clean.columns
):

    print("\nDelivery Feature Preview:")

    display(
        df_clean[
            [
                real_days,
                scheduled_days,
                "Delivery_Delay",
                "Delivery_Performance"
            ]
        ].head(10)
    )

    print("\nDelivery Performance Distribution:")

    display(
        df_clean[
            "Delivery_Performance"
        ].value_counts()
    )


# ============================================================
# 18. OUTLIER DETECTION USING IQR
# ============================================================

print("\n" + "=" * 70)
print("11. OUTLIER DETECTION")
print("=" * 70)

outlier_columns = [
    "Sales",
    "Shipping Cost",
    "Order Item Quantity",
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Delivery_Delay"
]

outlier_columns = [
    column
    for column in outlier_columns
    if column in df_clean.columns
]

outlier_summary = []

for column in outlier_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df_clean[column] < lower_bound)
        |
        (df_clean[column] > upper_bound)
    ).sum()

    outlier_summary.append({
        "Column": column,
        "Q1": round(Q1, 2),
        "Q3": round(Q3, 2),
        "IQR": round(IQR, 2),
        "Lower_Bound": round(lower_bound, 2),
        "Upper_Bound": round(upper_bound, 2),
        "Outlier_Count": int(outlier_count)
    })


outlier_table = pd.DataFrame(
    outlier_summary
)

display(outlier_table)


# ============================================================
# 19. OUTLIER VISUALIZATION
# ============================================================

print("\nOutlier Box Plots:")

for column in outlier_columns:

    plt.figure(figsize=(10, 4))

    sns.boxplot(
        x=df_clean[column]
    )

    plt.title(
        f"Outlier Detection - {column}"
    )

    plt.xlabel(column)

    plt.tight_layout()

    plt.show()


# ============================================================
# 20. OUTLIER TREATMENT EXPLANATION
# ============================================================

print("\n" + "=" * 70)
print("12. OUTLIER TREATMENT")
print("=" * 70)

print("""
Outliers were identified using the Interquartile Range (IQR)
method.

Outliers are not automatically deleted because extreme
logistics values may represent genuine business situations,
such as:

• High-value orders
• High shipping costs
• Long-distance shipments
• Unexpected delivery delays
• Large order quantities

Therefore, outliers are identified and reviewed rather than
blindly removed from the dataset.
""")


# ============================================================
# 21. MIN-MAX NORMALIZATION
# ============================================================

print("\n" + "=" * 70)
print("13. MIN-MAX NORMALIZATION")
print("=" * 70)

normalization_columns = [
    "Sales",
    "Shipping Cost",
    "Order Item Quantity",
    "Days for shipping (real)",
    "Days for shipment (scheduled)"
]

normalization_columns = [
    column
    for column in normalization_columns
    if column in df_clean.columns
]

if len(normalization_columns) > 0:

    minmax_scaler = MinMaxScaler()

    normalized_values = minmax_scaler.fit_transform(
        df_clean[normalization_columns]
    )

    normalized_df = pd.DataFrame(
        normalized_values,
        columns=[
            column + "_Normalized"
            for column in normalization_columns
        ],
        index=df_clean.index
    )

    df_clean = pd.concat(
        [
            df_clean,
            normalized_df
        ],
        axis=1
    )

    print(
        "✅ Min-Max normalization completed."
    )

    display(
        df_clean[
            [
                column + "_Normalized"
                for column in normalization_columns
            ]
        ].head()
    )


# ============================================================
# 22. STANDARDIZATION
# ============================================================

print("\n" + "=" * 70)
print("14. STANDARDIZATION")
print("=" * 70)

if len(normalization_columns) > 0:

    standard_scaler = StandardScaler()

    standardized_values = (
        standard_scaler.fit_transform(
            df_clean[normalization_columns]
        )
    )

    standardized_df = pd.DataFrame(
        standardized_values,
        columns=[
            column + "_Standardized"
            for column in normalization_columns
        ],
        index=df_clean.index
    )

    df_clean = pd.concat(
        [
            df_clean,
            standardized_df
        ],
        axis=1
    )

    print(
        "✅ Standardization completed."
    )

    display(
        df_clean[
            [
                column + "_Standardized"
                for column in normalization_columns
            ]
        ].head()
    )


# ============================================================
# 23. NORMALIZATION COMPARISON
# ============================================================

print("\n" + "=" * 70)
print("15. NORMALIZATION COMPARISON")
print("=" * 70)

if len(normalization_columns) > 0:

    comparison_column = normalization_columns[0]

    comparison = pd.DataFrame({

        "Original": df_clean[
            comparison_column
        ].head(10).values,

        "MinMax_Normalized": df_clean[
            comparison_column + "_Normalized"
        ].head(10).values,

        "Standardized": df_clean[
            comparison_column + "_Standardized"
        ].head(10).values
    })

    display(comparison)


# ============================================================
# 24. FINAL DATA QUALITY CHECK
# ============================================================

print("\n" + "=" * 70)
print("16. FINAL DATA QUALITY CHECK")
print("=" * 70)

print(
    "Original Rows:",
    original_rows
)

print(
    "Final Rows:",
    df_clean.shape[0]
)

print(
    "Original Columns:",
    original_columns
)

print(
    "Final Columns:",
    df_clean.shape[1]
)

print(
    "Total Missing Values:",
    df_clean.isnull().sum().sum()
)

print(
    "Total Duplicate Rows:",
    df_clean.duplicated().sum()
)


# ============================================================
# 25. BEFORE vs AFTER CLEANING
# ============================================================

print("\n" + "=" * 70)
print("17. BEFORE vs AFTER CLEANING")
print("=" * 70)

before_after = pd.DataFrame({

    "Metric": [
        "Rows",
        "Columns",
        "Missing Values",
        "Duplicate Rows"
    ],

    "Before Cleaning": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ],

    "After Cleaning": [
        df_clean.shape[0],
        df_clean.shape[1],
        df_clean.isnull().sum().sum(),
        df_clean.duplicated().sum()
    ]
})

display(before_after)


# ============================================================
# 26. FINAL DATASET PREVIEW
# ============================================================

print("\n" + "=" * 70)
print("18. FINAL PREPROCESSED DATASET")
print("=" * 70)

display(df_clean.head())

print("\nFinal Dataset Shape:")
print(df_clean.shape)


# ============================================================
# 27. SAVE PREPROCESSED DATASET
# ============================================================

print("\n" + "=" * 70)
print("19. SAVE PREPROCESSED DATASET")
print("=" * 70)

output_file = "DataCoSupplyChain_Preprocessed.csv"

df_clean.to_csv(
    output_file,
    index=False
)

print("✅ Preprocessed dataset saved successfully!")
print("Output file:", output_file)


# ============================================================
# 28. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("WEEK 2 TASK COMPLETED")
print("=" * 70)

print("""
DATA COLLECTION
✓ Public logistics dataset collected
✓ CSV dataset loaded successfully

DATA INSPECTION
✓ Dataset shape checked
✓ Columns inspected
✓ Data types checked
✓ Statistical summary generated

DATA CLEANING
✓ Duplicate records checked
✓ Duplicate records removed
✓ Missing values checked
✓ Missing values handled
✓ Date columns converted

FEATURE ENGINEERING
✓ Delivery_Delay created
✓ Delivery_Performance created

OUTLIER ANALYSIS
✓ IQR method used
✓ Outliers identified
✓ Box plots generated
✓ Extreme values reviewed

PREPROCESSING
✓ Min-Max normalization performed
✓ Standardization performed

VALIDATION
✓ Before vs After comparison generated
✓ Final data quality checked

OUTPUT
✓ DataCoSupplyChain_Preprocessed.csv created

The cleaned and preprocessed dataset is ready
for further logistics analysis and machine learning.
""")

print("=" * 70)
```
